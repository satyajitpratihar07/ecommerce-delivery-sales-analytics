```python
missing_values = df.isnull().sum()

print("Missing values:")
print(missing_values)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_table = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Percentage': missing_percentage
})

missing_table
```

```python
duplicates = df.duplicated().sum()

print("Duplicate rows:", duplicates)
```

```python
df = df.drop_duplicates()

print("Dataset after removing duplicates:")
print(df.shape)
```

```python
df['Order Date & Time'] = pd.to_datetime(
    df['Order Date & Time'],
    errors='coerce'
)

print(df['Order Date & Time'].dtype)
```

```python
invalid_dates = df['Order Date & Time'].isnull().sum()

print("Invalid dates:", invalid_dates)
```

```python
df['Order Date'] = df['Order Date & Time'].dt.date

df['Year'] = df['Order Date & Time'].dt.year
df['Month'] = df['Order Date & Time'].dt.month
df['Month Name'] = df['Order Date & Time'].dt.month_name()

df['Day'] = df['Order Date & Time'].dt.day
df['Day Name'] = df['Order Date & Time'].dt.day_name()

df['Hour'] = df['Order Date & Time'].dt.hour

print("Date columns created successfully!")
```

```python
total_orders = df['Order ID'].nunique()

print("Total Orders:", total_orders)
```

```python
total_sales = df['Order Value (INR)'].sum()

print(f"Total Sales: ₹{total_sales:,.2f}")
```

```python
average_order_value = df['Order Value (INR)'].mean()

print(f"Average Order Value: ₹{average_order_value:,.2f}")
```

```python
median_order_value = df['Order Value (INR)'].median()

print(f"Median Order Value: ₹{median_order_value:,.2f}")
```

```python
total_customers = df['Customer ID'].nunique()

print("Total Customers:", total_customers)
```

```python
average_delivery_time = df['Delivery Time (Minutes)'].mean()

print(f"Average Delivery Time: {average_delivery_time:.2f} minutes")
```

```python
average_rating = df['Service Rating'].mean()

print(f"Average Service Rating: {average_rating:.2f}")
```

```python
best_platform = platform_analysis['Total_Sales'].idxmax()
best_category = category_analysis['Total_Sales'].idxmax()
best_rating_platform = platform_analysis['Average_Rating'].idxmax()
fastest_platform = platform_analysis['Average_Delivery_Time'].idxmin()

delay_rate = (
    (df['Delivery Delay'] == 'Yes').mean() * 100
)

refund_rate = (
    (df['Refund Requested'] == 'Yes').mean() * 100
)

print("========== BUSINESS SUMMARY ==========")

print(f"Total Orders              : {total_orders:,}")
print(f"Total Customers           : {total_customers:,}")
print(f"Total Sales               : ₹{total_sales:,.2f}")
print(f"Average Order Value       : ₹{average_order_value:,.2f}")
print(f"Average Delivery Time     : {average_delivery_time:.2f} minutes")
print(f"Average Rating            : {average_rating:.2f}")
print(f"Best Platform by Sales    : {best_platform}")
print(f"Best Category by Sales    : {best_category}")
print(f"Best Rated Platform       : {best_rating_platform}")
print(f"Fastest Platform          : {fastest_platform}")
print(f"Delivery Delay Rate       : {delay_rate:.2f}%")
print(f"Refund Request Rate       : {refund_rate:.2f}%")
```

```python
final_analysis = pd.DataFrame({
    'Metric': [
        'Total Orders',
        'Total Customers',
        'Total Sales',
        'Average Order Value',
        'Average Delivery Time',
        'Average Rating',
        'Delivery Delay Rate',
        'Refund Request Rate',
        'Best Platform',
        'Best Category',
        'Best Rated Platform',
        'Fastest Platform'
    ],

    'Result': [
        f"{total_orders:,}",
        f"{total_customers:,}",
        f"₹{total_sales:,.2f}",
        f"₹{average_order_value:,.2f}",
        f"{average_delivery_time:.2f} minutes",
        f"{average_rating:.2f}",
        f"{delay_rate:.2f}%",
        f"{refund_rate:.2f}%",
        best_platform,
        best_category,
        best_rating_platform,
        fastest_platform
    ]
})

final_analysis
```
